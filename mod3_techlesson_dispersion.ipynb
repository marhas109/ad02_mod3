## Overview ##

In this technical lesson, we will walk through calculating the standard deviation, mean absolute deviation, and IQR on a data sample using Python. We will also go over how to use numpy's methods to extract these measures without having to calculate them manually.

#### Situation 1: Characteristic Range for Flower Petal Lengths ####

Let's revisit our list of petal lengths for members of the Iris Virginica subspecies. In a previous lesson, we computed measures of central tendency for this data (i.e., the mean and median). This data was symmetrically distributed, with no outliers and a clear central tendency. 

In this case, all the measures of dispersion (standard deviation, MAD, and IQR) should be on the same scale. However, because they measure different things, it is unreasonable to expect them to be the same. Let's go ahead and calculate all of these.

**Variance**

```python
virginica_petal_length = [6.0, 5.1, 5.9, 5.6, 5.8, 6.6, 4.5, 6.3, 5.8, 6.1, 5.1, 5.3, 5.5, 5.0, 5.1, 5.3, 5.5, 6.7, 6.9, 5.0, 5.7, 4.9, 6.7, 4.9, 5.7, 6.0, 4.8, 4.9, 5.6, 5.8, 6.1, 6.4, 5.6, 5.1,
5.6, 6.1, 5.6, 5.5, 4.8, 5.4, 5.6, 5.1, 5.1, 5.9, 5.7, 5.2, 5.0, 5.2, 5.4, 5.1]
```

```python
def get_variance(sample):

    # calculate the sample mean
    sample_mean = sum(sample)/len(sample)
   
    sum_of_squares = 0
    for length in sample:
        # calculate the sum of squares by subtracting the sample mean
        # from each length, squaring the result, and adding it to the total
        sum_of_squares += (length - sample_mean)**2
       
    # divide the sum of squares by the number of items in the sample -1 to calculate variance
    variance = sum_of_squares/(len(sample)- 1)
   
    return variance
```

```python
get_variance(virginica_petal_length)
```

```python
# variance with numpy

import numpy as np

virginica_array = np.array(virginica_petal_length)
virginica_array.var(ddof = 1)
```

**Standard Deviation**

```python
def get_stddev(sample):

    stdev = (get_variance(sample))**0.5

    return stdev
```

```python
get_stddev(virginica_petal_length)
```

```python
# using numpy

virginica_array.std(ddof = 1)
```

**Mean Absolute Deviation (MAD)**

```python
def get_mad(sample):

    # First, calculate the sample mean
    sample_mean = sum(sample)/len(sample)
   
    sum_of_devs = 0
    for length in sample:
        # find deviation by subtracting the sample mean
        # ftake the absolute value of the deviation and adding it to the total
        sum_of_devs += abs(length - sample_mean)
       
    # Divide the sum of deviations by the number of items in the sample to calculate MAD  
    mad = sum_of_devs/(len(sample))
   
    return mad
```

```python
get_mad(virginica_petal_length)
```

**Interquartile Range (IQR)**

Now let's calculate the IQR. In principle, we could find a function that sorts the data, splits it into quarters, and then extracts the range of the middle half. We are not going to take this route here. Instead, we will show you two methods to extract the IQR using *vmpy* and *scipy*.

```python
# using vmpy to calculate percentiles
Q1 = np.percentile(virginica_petal_length, 25)
Q3 = np.percentile(virginica_petal_length, 75)
IQR = Q3 - Q1

print(Q1)
print(Q3)
print(IQR)
```

```python
# using scipy
from scipy.stats import iqr
iqr(virginica_petal_length)
```

**Mean Absolute Deviation**

```python
def compute_median(data):
  """
  This function calculates the median of a list of numbers.
  Args:
      data: A list of numbers.
  Returns:
      The median of the data.
  """

  # Sort the data in ascending order.
  sorted_data = sorted(data)

  # Calculate the length of the data.
  n = len(sorted_data)

  # If the length is odd, the median is the middle element.
  if n % 2 == 1:
    return sorted_data[n // 2]

  # If the length is even, the median is the average of the two middle elements.
  else:
    middle_index1 = n // 2
    middle_index2 = middle_index1 - 1
    return (sorted_data[middle_index1] + sorted_data[middle_index2]) / 2
```

```python
compute_median(weekly_sales)
```

**Interquartile Range**

```python
iqr(weekly_sales)
```
